```python
%sql
-- Databricks notebook source
-- MAGIC %md
-- MAGIC # 00 - Workspace & Unity Catalog Setup
-- MAGIC **Purpose:** Initialize the `ecom_poc` schema and managed `raw_files` volume in an existing writable Unity Catalog[cite: 4, 5].
-- MAGIC 
-- MAGIC **Instructions:** Replace `<YOUR_UC_CATALOG>` below with a writable catalog visible in your Catalog Explorer (e.g., `workspace` or your trial catalog name) where you have `CREATE SCHEMA` privileges[cite: 4].

-- COMMAND ----------

-- MAGIC %md
-- MAGIC ### 1. Inspect Available Catalogs & Current User
-- MAGIC Check which catalogs exist and verify your current workspace identity[cite: 4].

-- COMMAND ----------

SELECT current_user() AS active_user, current_metastore() AS active_metastore;
SHOW CATALOGS;

-- COMMAND ----------

-- MAGIC %md
-- MAGIC ### 2. Configure Catalog, Schema, and Managed Volume
-- MAGIC Execute the suggested setup commands to create the landing zone `/Volumes/<YOUR_UC_CATALOG>/ecom_poc/raw_files/`[cite: 4].

-- COMMAND ----------

-- Replace <YOUR_UC_CATALOG> with your actual writable catalog name
USE CATALOG workspace;

CREATE SCHEMA IF NOT EXISTS ecom_poc
COMMENT 'E-commerce Medallion Lakehouse POC schema containing Bronze, Silver, and Gold layers';

USE SCHEMA ecom_poc;

CREATE VOLUME IF NOT EXISTS raw_files
COMMENT 'Landing zone for raw, immutable source CSV extracts (orders.csv and transactions.csv)';

-- COMMAND ----------

-- MAGIC %md
-- MAGIC ### 3. Verify Governance Objects & Volume Path
-- MAGIC Confirm that the schema and managed volume were created properly[cite: 5]. Take a screenshot of these outputs and Catalog Explorer for your submission evidence[cite: 5].

-- COMMAND ----------

DESCRIBE SCHEMA EXTENDED ecom_poc;
DESCRIBE VOLUME raw_files;

-- COMMAND ----------

-- MAGIC %md
-- MAGIC ### 4. Check Uploaded Files (Run After Uploading CSVs)
-- MAGIC Once you manually upload `orders.csv` and `transactions.csv` via **New → Add or upload data → Upload files to a volume** into `/Volumes/<YOUR_UC_CATALOG>/ecom_poc/raw_files/`, run this cell to verify the file paths and byte sizes.

-- COMMAND ----------

LIST '/Volumes/<YOUR_UC_CATALOG>/ecom_poc/raw_files/';
```
